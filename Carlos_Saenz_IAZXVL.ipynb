Create a House class that represents a house; attributes:


*   area
*   number of rooms
*   number of bathrooms
*   number of floors
*   has garage
*   has garden
*   any other attributes that you think is important



instance methods:

*   price calculation

There should be at least 1 default parameter

(4 points)

```python
class House():
  def __init__(self, area, n_rooms, n_bathrooms, n_floors, n_garage=1, garden=False, age=0):
    self.area = area
    self.rooms = n_rooms
    self.bathrooms = n_bathrooms
    self.floors = n_floors
    self.n_garage = n_garage
    self.garden = garden
    self.age = age

  def price(self):
    price = (self.area*1000 + self.rooms*5000 + self.bathrooms*2000 + self.n_garage*3500) - self.age*3000

    if self.garden:
      price += 7000

    return price
```

```python
house = House(120, 4, 2, 2, 3, False, 5)

print(house.price())
```

```python
class Person():
  def __init__(self, name, age, married, n_children, n_cars, money):
    self.name = name
    self.age = age
    self.married = married
    self.children = n_children
    self.cars = n_cars
    self.money = money

    if self.married:
          self.family_members = 2 + self.children
    else:
          self.family_members = 1 + self.children
```

###A) (5 points)
 Create an Animal class and choose 3 animals that have some things in common (characteristics and behaviour)!

Write the Parent class and the derived classes of the species!

The Child classes should inherit some attributes and override some methods of the Parent class, but should also extend the state with own attributes in each subclass and extend the behaviour with own methods.

Decide yourself:

which methods belong to parent


which should be overridden


which attributes are shared. Explain your choice.

```python
class Animal:
    def __init__(self, specie, habitat, diet):
        self.specie = specie
        self.habitat = habitat
        self.diet = diet

    def eat(self):
        print(f"The {self.specie} is eating {self.diet}")

    def sleep(self):
        print(f"The {self.specie} is sleeping in {self.habitat}")

class Monkeys(Animal):
    def __init__(self, specie, habitat, diet, name, bananas=0):
        super().__init__(specie, habitat, diet)
        self.name = name
        self.bananas = bananas
    def monkey_info(self):
        print(f"{self.name} is a {self.specie}, he lives in {self.habitat} and eats {self.diet}")
    def get_banana(self):
        self.bananas += 1
        print(f"{self.name} has {self.bananas} bananas")

class Lions(Animal):
    def __init__(self, specie, habitat, diet, hunt=True, captured_prey=0 ):
        super().__init__(specie, habitat, diet)
        self.hunt = hunt
        self.captured_prey = captured_prey
    def lion_info(self):
        print(f"The {self.specie} lives in {self.habitat} and eats {self.diet}")
    def get_prey(self):
      if self.hunt:
        print(f"The {self.specie} is hunting")
        self.captured_prey += 1
        print(f"The {self.specie} has {self.captured_prey} prey")
      else:
        print(f"The {self.specie} is not hunting")

class Crocodiles(Animal):
    def __init__(self, specie, habitat, diet, name, age):
        super().__init__(specie, habitat, diet)
        self.age = age
        self.name = name
    def crocodile_info(self):
        print(f"{self.name} is a {self.specie} and lives in {self.habitat} and eats {self.diet}")
    def get_age(self):
        print(f"{self.name} is {self.age} years old")
    def crocodile_birthday(self):
        self.age += 1
        print(f"{self.name} is now {self.age} years old")



```


Explanation of my choice:
I created a parent class called Animal because monkeys, lions, and crocodiles are all animals and share common characteristics. For this reason, the shared attributes are specie, habitat, and diet. These attributes belong to the parent class because every animal has a species, lives in a habitat, and has a type of food.
The methods eat() and sleep() also belong to the parent class because they are common behaviours for all animals.
Then, I created three child classes: Monkeys, Lions, and Crocodiles. These classes inherit the shared attributes and methods from Animal, but each one also has its own attributes and behaviours.
The child classes also extend the behaviour of the parent class with their own information methods, such as monkey_info(), lion_info(), and crocodile_info(). This shows inheritance because the child classes reuse common animal information, but also specialization because each animal has unique attributes and methods.



```python
monkey_1 = Monkeys("Monkey", "Jungle", "fruits", "Jorge")
monkey_1.monkey_info()
monkey_1.eat()
monkey_1.sleep()
# Give bananas
monkey_1.get_banana()
monkey_1.get_banana()

lion_1 = Lions("Lion", "Savannah", "meat", True)
lion_1.lion_info()
lion_1.eat()
lion_1.sleep()
# Get prey
lion_1.get_prey()
lion_1.get_prey()

crocodile_1 = Crocodiles("Crocodile", "River", "meat", "Croky", 25)
# Get age
crocodile_1.crocodile_info()
crocodile_1.get_age()
crocodile_1.crocodile_birthday()
crocodile_1.sleep()
crocodile_1.eat()
```

###B) (6 points)

Create classes and use inheritance to calculate the area of different shapes (specifically, circle and square, you can add triangle or anything else if you want).

Requirements:


1.	Create a base class called Shape that has a method named area().

This method should not have any implementation
2. Create at least two subclasses:

Circle — which takes a radius as a parameter

Square — which takes a side as a parameter

write the area-calculating function for both (or for all)

3. Create objects of both Circle and Square, and print their areas.

Extra: Store these objects in a list and use a loop to print the area of each shape.

```python
import math

class Shape:
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius ** 2

class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side ** 2

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    def area(self):
        return (self.base * self.height) / 2
```

```python
from typing import Self
circle_1 = Circle(5)
square_1 = Square(4)
triangle_1 = Triangle(6, 3)

print("Circle area:", circle_1.area())
print("Square area:", square_1.area())
print("Triangle area:", triangle_1.area())


# Extra:
shapes = [("circle_1", circle_1), ("square_1", square_1), ("triangle_1", triangle_1)]
for name, s in shapes:
    print(f"Area of {name}: {s.area()}")
```

Combine two lists into tuples

```python
names = ["Anna", "Bob", "Charlie"]
scores = [85, 92, 78]
```

Expected output: [('Anna', 85), ('Bob', 92), ('Charlie', 78)]

```python
conbined_list = []
for n, s in zip(names, scores):
  conbined_list.append((n, s))
print(conbined_list)
```

Count character frequency

```python
text = "banana"
```

Expected output: {'b': 1, 'a': 3, 'n': 2}

```python
charac = []
for c in text:
  if c not in charac:
    charac.append(c)
print(charac)

charac_freq = {}
for c in charac:
  charac_freq[c] = text.count(c)
print(charac_freq)
```
